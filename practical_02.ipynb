```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 

data = {
    'Student':['A','B','C','D','E'],
    'Marks':[85,75,90,65,85],
    'Study_hours':[6,5,8,4,7]
}

df = pd.DataFrame(data)
print(data)
print(df)

#line chart
plt.figure(figsize=(5,4))
plt.plot(df['Student'],df['Marks'],marker='o',)
plt.title("Line Chart")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()

#Bar chart
plt.figure(figsize=(5,4))
plt.bar(df['Student'],df['Marks'])
plt.title("Bar Chart")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()

#Histogram
plt.figure(figsize=(5,4))
plt.hist(df['Marks'])
plt.title("Histogram")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()

#scatter plot
plt.figure(figsize=(5,4))
plt.scatter(df['Study_hours'],df['Marks'])
plt.title("Scatter Plot")
plt.xlabel("study hours")
plt.ylabel("Marks")
plt.show()

#box plot
plt.figure(figsize=(5,4))
plt.boxplot(df['Marks'])
plt.title("Box Plot")
plt.xlabel("Marks")
plt.ylabel("Values")
plt.show()

# Heatmap
plt.figure(figsize=(5,4))
corr_matrix = df[['Marks', 'Study_hours']].corr()   # Select only numeric columns
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title("Heatmap")
plt.show()
```
