```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
data = {
    'Student':['A','B','C','D','E'],
    'Marks':[85,75,90,65,85],
    'Study_hours':[6,5,8,4,7]
}
```

```python
df = pd.DataFrame(data)
print(data)
print(df)
```

```python
#line chart

plt.figure(figsize=(5,4))
plt.plot(df['Student'],df['Marks'],marker='o',)
plt.title("Line Chart")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
#Bar chart

plt.figure(figsize=(5,4))
plt.bar(df['Student'],df['Marks'])
plt.title("Bar Chart")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
#Histogram

plt.figure(figsize=(5,4))
plt.hist(df['Student'])
plt.title("Histogram")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
#scatter chart

plt.figure(figsize=(5,4))
plt.scatter(df['Student'],df['Marks'],marker='o',)
plt.title("Line Chart")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```
