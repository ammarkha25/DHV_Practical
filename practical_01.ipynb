```python
import numpy as np
import statistics as st
```

```python
data = [12,15,18,20,22,25,18,20,30,35]
print("Dataset:", data)
print("Mean:", np.mean(data))
print("Median:", np.median(data))
print("Mode:", st.mode(data))
print("Standard Deviation:", np.std(data))
print("Variance:", np.var(data))
```
